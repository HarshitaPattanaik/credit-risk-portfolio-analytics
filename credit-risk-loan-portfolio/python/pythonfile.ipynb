```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Upload your CSV file to Google Colab
from google.colab import files
uploaded = files.upload()
```

```python
# 2. Load the CSV into a Pandas DataFrame
# (Replace 'portfolio_risk_metrics.csv' with your exact uploaded filename if different)
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)
```

```python
# 3. Explore the dataset structure and basic statistics
print("--- Dataset Info ---")
display(df.info())
```

```python
print("\n--- Summary Statistics ---")
display(df.describe())
```

```python
# Preview the first few rows
display(df.head())
```

```python
# 4. Statistical Visualization: Outstanding Balance Distribution by Delinquency & Risk Grade
plt.figure(figsize=(12, 6))
sns.set_theme(style="whitegrid")
```

```python
# Create a bar plot breaking down exposure by delinquency buckets and risk grades
ax = sns.barplot(
    data=df,
    x='delinquency_bucket',
    y='total_outstanding_balance',
    hue='risk_grade',
    palette='muted'
)

plt.title('Portfolio Outstanding Balance Exposure by Delinquency & Risk Grade', fontsize=14, fontweight='bold')
plt.xlabel('Delinquency Bucket (DPD)', fontsize=12)
plt.ylabel('Total Outstanding Balance ($)', fontsize=12)
plt.xticks(rotation=15)
plt.legend(title='Risk Grade', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

# Display the plot
plt.show()
```
